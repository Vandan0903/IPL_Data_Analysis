```python
import pandas as pd
```

```python
matches = pd.read_csv("matches.csv")
deliveries = pd.read_csv("deliveries.csv")
```

```python
matches.head(5)
```

```python
deliveries.head(5)
```

```python
matches.info()
deliveries.info()
```

```python
matches.isnull().sum()
```

```python
deliveries.isnull().sum()
```

```python
matches['winner'] = matches['winner'].fillna('No Result')
matches['player_of_match'] = matches['player_of_match'].fillna('Unknown')
```

```python
matches.isnull().sum()
```

```python
matches['date'] = pd.to_datetime(matches['date'])
```

```python
matches.drop_duplicates(inplace=True)
deliveries.drop_duplicates(inplace=True)
```

```python
matches['city'] = matches['city'].fillna('Unknown')
matches['result_margin'] = matches['result_margin'].fillna('No Result')
```

```python
matches.isnull().sum()
```

```python
deliveries['player_dismissed'] = deliveries['player_dismissed'].fillna('Not Out')
```

```python
deliveries['extras_type'] = deliveries['extras_type'].fillna('None')
deliveries['dismissal_kind'] = deliveries['dismissal_kind'].fillna('None')
```

```python
deliveries.isnull().sum()
```

```python
deliveries['fielder'] = deliveries['fielder'].fillna('No Fielder')
```

```python
deliveries.isnull().sum()
```

```python
matches['method'] = matches['method'].fillna('Normal Match')
```

```python
matches.isnull().sum()
```

```python
matches['target_overs'] = matches['target_overs'].fillna(0)
```

```python
matches.isnull().sum()
```

```python
matches['target_runs'] = matches['target_runs'].fillna(0)
```

```python
matches.isnull().sum()
```

```python
matches['win_type'] = matches['result'].apply(
    lambda x: 'Defend' if x == 'runs' else ('Chase' if x == 'wickets' else 'No Result')
)
```

```python
matches[['result', 'win_type']].head(5)
```

```python
matches['year'] = matches['date'].dt.year
```

```python
matches['toss_win_match_win'] = (
    matches['toss_winner'] == matches['winner']
)
```

```python
matches['toss_win_match_win'].head(5)
```

```python
total_runs = deliveries.groupby('match_id')['total_runs'].sum().reset_index()
wickets = deliveries[deliveries['is_wicket'] == 1] \
    .groupby('match_id').size().reset_index(name='total_wickets')
```

```python
df = matches.merge(total_runs, left_on='id', right_on='match_id')
df = df.merge(wickets, left_on='id', right_on='match_id')
```

```python
df.tail(5)
```

```python
df.drop(columns=['match_id_x', 'match_id_y'], inplace=True)
```

```python
df.head(5)
```

```python
runs_innings = deliveries.groupby(['match_id', 'inning'])['total_runs'].sum().reset_index()
runs_pivot = runs_innings.pivot(index='match_id',  columns='inning',  values='total_runs').reset_index()
runs_pivot = runs_pivot.rename(columns={ 1: 'runs_1st_innings', 2: 'runs_2nd_innings'})
runs_pivot[['runs_1st_innings', 'runs_2nd_innings']] = runs_pivot[['runs_1st_innings', 'runs_2nd_innings']].fillna(0)

   
```

```python
wickets_innings = deliveries[deliveries['is_wicket'] == 1].groupby(['match_id', 'inning']).size().reset_index(name='wickets')
wickets_pivot = wickets_innings.pivot( index='match_id', columns='inning', values='wickets').reset_index()
wickets_pivot = wickets_pivot.rename(columns={ 1: 'wickets_1st_inning', 2: 'wickets_2nd_inning'})  
wickets_pivot[['wickets_1st_inning', 'wickets_2nd_inning']] = wickets_pivot[['wickets_1st_inning', 'wickets_2nd_inning']].fillna(0) 

   

 
```

```python
df = df.merge(runs_pivot, left_on='id', right_on='match_id')
df = df.merge(wickets_pivot, left_on='id', right_on='match_id')
```

```python
df.head()
```

```python
df.drop(columns=['match_id_x', 'match_id_y'], inplace=True, errors='ignore')
```

```python
df.head()
```

```python
runs_pivot = runs_pivot.rename(columns={1: 'runs_1st_innings', 2: 'runs_2nd_innings'})

for col in ['runs_1st_innings', 'runs_2nd_innings']:
    if col not in runs_pivot.columns:
        runs_pivot[col] = 0

runs_pivot[['runs_1st_innings', 'runs_2nd_innings']] = runs_pivot[['runs_1st_innings', 'runs_2nd_innings']].fillna(0)
```

```python
df.head(5)
```

```python
df = df.drop(columns=[
    col for col in df.columns 
    if '_x' in str(col) or '_y' in str(col) or col in [3,4,5,6]
], errors='ignore')
```

```python
df.head(5)
```

```python
df[['runs_1st_innings','runs_2nd_innings','wickets_1st_inning','wickets_2nd_inning']].isnull().sum()
```

```python
df[['runs_1st_innings','runs_2nd_innings','wickets_1st_inning','wickets_2nd_inning']] = \
df[['runs_1st_innings','runs_2nd_innings','wickets_1st_inning','wickets_2nd_inning']].astype(int)
```
